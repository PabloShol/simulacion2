```python
import numpy as np

# Matriz de transición P
P = np.array([
    [0.1, 0.2, 0.4, 0.3],
    [0.4, 0.0, 0.4, 0.2],
    [0.3, 0.3, 0.0, 0.4],
    [0.2, 0.1, 0.4, 0.3]
    ])

# Planteamiento: pi * (P - I) = 0  =>  (P.T - I) * pi^T = 0
# Reemplazamos la última fila con la condición de suma: sum(pi) = 1
A = (P.T - np.eye(4))
A[-1] = np.ones(4)
b = np.array([0, 0, 0, 1])
# Resolver el sistema lineal
pi = np.linalg.solve(A, b)

estados = ["Aerobics (x)", "Masaje (y)", "Pesas (z)", "Yoga (w)"]
print("Distribución límite (estacionaria):")
for estado, prob in zip(estados, pi):
      print(f"  {estado}: {prob:.4f}")
```
